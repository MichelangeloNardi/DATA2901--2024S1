# Income

```python
# Import necessary Libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Reading in the Data into a DataFrame, as it has no geometry info
relative_path = "Income.csv"
income_data = pd.read_csv(relative_path)
income_index = list(income_data['sa2_code21'])
income_data.set_index('sa2_code21',inplace=True)
income_data.head(5)
```

The following are numerous checks for different types of null values (Null, Na, Non-Numeric, Negative)

```python
# Initial Test for null values
for col in list(income_data.columns)[2:6]: #This selects earners, median_age, median_income and mean_income
    size = len(income_data[income_data[col].isnull()]) #Number of Null values
    size += len(income_data[income_data[col].isna()]) #Number of NA values
    print(size)
```

```python
#This code finds all entries which are non-numeric in the columns we expect to have numbers
problem_rows = [] #Keep a list of all rows with null values
for row in income_index:
    for col in list(income_data.columns)[2:6]: #This selects earners, median_age, median_income and mean_income
        try:
            income_data[col][row] = pd.to_numeric(income_data[col][row]) #Tries to make it a number, if it fails it prints the line
            if income_data[col][row] < 0: #Tests for negative numbers which are also an irregularity
                problem_rows.append(row)
            break
        except:
            problem_rows.append(row)
            break
income_data.loc[problem_rows]
```

# Comparisons to Population

Note there is a difference in size between this data set and the population dataset, and so by importing that dataset too we can check which ones are missing.

```python
relative_path = "Population.csv"
population_data = pd.read_csv(relative_path)
```

```python
pop_codes = list(population_data['sa2_code'])
inc_codes = list(income_data['sa2_code21'])
all_codes = set(pop_codes + inc_codes)
len(all_codes)

missing_codes = list(set(all_codes).difference(set(pop_codes)))
missing_codes
```
