```python
#Reading in the Data into a DataFrame, as it has no geometry info
relative_path = "Population.csv"
population_data = pd.read_csv(relative_path)
population_data.head(3)
```

```python
sum(population_data.total_people)
```

```python
#Create a list of all age range column names
age_ranges = []
for num in range(17):
    age_range = str(5*num) + '-' + str(5*num+4) + '_people'
    age_ranges.append(age_range)
age_ranges.append('85-and-over_people')
print(age_ranges)
```

The Following code pieces are different tests for the presence of potential null values throughout the columns, all of which show that there are no null values.

```python
#Check for explicit Null values in each column
for age in age_ranges:
    print(f'{age}, Null Values: {len(population_data[population_data[age].isnull()])}')
```

```python
#Check for negative entries
for age in age_ranges:
    print(f'{age}, Negative Values: {len(population_data[population_data[age] < 0])}')
```

The next test is to determine whether all the sums of population for each area is correct, and the result of an empty dataframe means that it is correct.

```python
# Calculate the total population from each row
calc_population = population_data[age_ranges].sum(axis=1, numeric_only=True) #Using population_data[age_ranges] to remove sa2_code which would otherwise be added.
population_data['total_people_calc'] = calc_population
population_data[population_data['total_people'] != population_data['total_people_calc']]
```
